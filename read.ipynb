# 🏦 Loan Approval Prediction System

A machine learning web application that predicts whether a loan application will be approved or rejected.

## 🎯 Features
- Decision Tree Classifier with 97.8% accuracy
- SMOTE for handling imbalanced data
- Interactive Streamlit interface
- Real-time predictions

## 📊 Model Performance
- **Accuracy**: 97.8%
- **Precision**: 98.5%
- **Recall**: 97.9%
- **F1-Score**: 98.2%

## 🛠️ Tech Stack
- Python
- Scikit-learn
- Streamlit
- Pandas / NumPy
- Imbalanced-learn (SMOTE)

## 🚀 How to Run Locally
\`\`\`bash
pip install -r requirements.txt
streamlit run app.py
\`\`\`

## 📁 Files
- `app.py` - Streamlit application
- `data train.ipynb` - Model training notebook
- `model.pkl` - Trained Decision Tree model
- `scaler.pkl` - StandardScaler
- `encoders.pkl` - Label encoders
